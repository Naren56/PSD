```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Baca dua gambar hasil dari KNIME
img1 = mpimg.imread("Data/Bar_Chart_KNIME.png")
img2 = mpimg.imread("Data/Scatter_Plot_KNIME.png")

# Buat figure lebih besar (width=12 inch, height=14 inch)
plt.figure(figsize=(12, 14))

# Gambar pertama (atas)
plt.subplot(2, 1, 1)
plt.imshow(img1)
plt.axis("off")  # sembunyikan sumbu
plt.title("Bar Chart dari KNIME", fontsize=18, fontweight="bold", pad=20)

# Gambar kedua (bawah)
plt.subplot(2, 1, 2)
plt.imshow(img2)
plt.axis("off")
plt.title("Scatter Plot dari KNIME", fontsize=18, fontweight="bold", pad=20)

# Judul utama di atas semua gambar
plt.suptitle("Hasil Visualisasi dari KNIME", fontsize=20, fontweight="bold", y=0.99)

# Atur tata letak agar proporsional
plt.tight_layout(rect=[0, 0, 1, 0.96])  # biar tidak nabrak judul
plt.show()
```
